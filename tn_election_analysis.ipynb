```python
import pandas as pd
import numpy as np
```

```python
df2021 = pd.read_csv("tn_2021_results.csv")
df2026 = pd.read_csv("tn_2026_results.csv")
master = pd.read_csv("constituency_master.csv")
```

```python
winner2021 = df2021.loc[df2021.groupby('ac_number')['votes'].idxmax()]
winner2026 = df2026.loc[df2026.groupby('ac_number')['votes'].idxmax()]
```

```python
winner2021.head()
```

```python
winner2026.head()
```

```python
print(winner2021.shape)
print(winner2026.shape)
```

```python
sorted2021 = df2021.sort_values(['ac_number', 'votes'], ascending=[True, False])
sorted2026 = df2026.sort_values(['ac_number', 'votes'], ascending=[True, False])
```

```python
top2_2021 = sorted2021.groupby('ac_number').head(2)
top2_2026 = sorted2026.groupby('ac_number').head(2)
```

```python
margin2021 = top2_2021.groupby('ac_number').agg({
    'votes': lambda x: x.iloc[0] - x.iloc[1]
}).reset_index()

margin2021.columns = ['ac_number', 'margin']
```

```python
margin2026 = top2_2026.groupby('ac_number').agg({
    'votes': lambda x: x.iloc[0] - x.iloc[1]
}).reset_index()

margin2026.columns = ['ac_number', 'margin']
```

```python
winner2021 = winner2021.merge(margin2021, on='ac_number')
winner2026 = winner2026.merge(margin2026, on='ac_number')
```

```python
winner2021[['constituency', 'candidate', 'party', 'votes', 'margin']].head()
```

```python
closest_battles_2026 = winner2026.sort_values('margin').head(10)

closest_battles_2026[['constituency', 'candidate', 'party', 'margin']]
```

```python
w21 = winner2021[['ac_number', 'constituency', 'party']]
w26 = winner2026[['ac_number', 'constituency', 'party']]
```

```python
w21 = w21.rename(columns={'party': 'party_2021'})
w26 = w26.rename(columns={'party': 'party_2026'})
```

```python
flip_analysis = w21.merge(w26, on=['ac_number', 'constituency'])
```

```python
flip_analysis['flipped'] = flip_analysis['party_2021'] != flip_analysis['party_2026']
```

```python
flip_analysis['flipped'].value_counts()
```

```python
flipped_seats = flip_analysis[flip_analysis['flipped'] == True]

flipped_seats.head(20)
```

```python
party_flips = flipped_seats.groupby(
    ['party_2021', 'party_2026']
).size().reset_index(name='seats_changed')

party_flips.sort_values('seats_changed', ascending=False)
```

```python
winner2021 = winner2021.merge(
    master[['ac_number', 'region']],
    on='ac_number',
    how='left'
)

winner2026 = winner2026.merge(
    master[['ac_number', 'region']],
    on='ac_number',
    how='left'
)
```

```python
winner2021 = winner2021.drop(columns=['region_y'])
winner2021 = winner2021.rename(columns={'region_x': 'region'})

winner2026 = winner2026.drop(columns=['region_y'])
winner2026 = winner2026.rename(columns={'region_x': 'region'})
```

```python
winner2021[['constituency', 'party', 'region']].head()
```

```python
region_2021 = winner2021.groupby(
    ['region', 'party']
).size().reset_index(name='seats_2021')

region_2021.head()
```

```python
region_2026 = winner2026.groupby(
    ['region', 'party']
).size().reset_index(name='seats_2026')

region_2026.head()
```

```python
regional_shift = region_2021.merge(
    region_2026,
    on=['region', 'party'],
    how='outer'
).fillna(0)
```

```python
regional_shift['seat_change'] = (
    regional_shift['seats_2026']
    - regional_shift['seats_2021']
)
```

```python
regional_shift.sort_values(
    'seat_change',
    ascending=False
).head(20)
```
